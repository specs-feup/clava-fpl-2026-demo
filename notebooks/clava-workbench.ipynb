# Clava notebook workbench

This notebook calls the npm project in `../clava-scripts` and then inspects generated files.

```python
from pathlib import Path
import subprocess

NOTEBOOK_DIR = Path.cwd()
CLAVA_SCRIPTS_DIR = (NOTEBOOK_DIR / '..' / 'clava-scripts').resolve()
CLAVA_SCRIPTS_DIR
```

```python
subprocess.run(['npm', 'install'], cwd=CLAVA_SCRIPTS_DIR, check=True)
subprocess.run(['npm', 'run', 'build'], cwd=CLAVA_SCRIPTS_DIR, check=True)
subprocess.run(['npm', 'run', 'run'], cwd=CLAVA_SCRIPTS_DIR, check=True)
```

```python
generated = list((CLAVA_SCRIPTS_DIR / 'woven_code').glob('*')) if (CLAVA_SCRIPTS_DIR / 'woven_code').exists() else []
generated
```
